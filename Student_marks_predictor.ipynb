```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Study_Hours': [1, 2, 2.5, 3, 3.5, 4, 5, 5.5, 6, 7, 8, 9],
    'Marks': [35, 42, 45, 50, 55, 60, 68, 70, 75, 82, 88, 94]
}

df = pd.DataFrame(data)

df

```

```python
plt.scatter(df['Study_Hours'], df['Marks'])

plt.title('Study Hours vs Marks')
plt.xlabel('Study Hours')
plt.ylabel('Marks')

plt.show()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
X = df[['Study_Hours']]
```

```python
y = df[ 'Marks']
```

```python
plt.scatter(df['Study_Hours'], df['Marks'])
```

```python
model = LinearRegression()
```

```python
model.fit(X, y)
```

```python
prediction = model.predict([[6.5]])

print("Predicted Marks:", prediction[0])
```

```python
plt.scatter(df['Study_Hours'], df['Marks'])
plt.scatter(6.5, prediction[0], s=100)

plt.title('Study Hours vs Marks — Prediction')
plt.xlabel('Study Hours')
plt.ylabel('Marks')

plt.show()
```

```python
from sklearn.metrics import r2_score
```

```python
predicted_marks = model.predict(X)

r2 = r2_score(y, predicted_marks)

print("R² Score:", r2)
```

```python
plt.scatter(df['Study_Hours'], df['Marks'])

plt.plot(df['Study_Hours'], model.predict(X))

plt.title('Student Marks Predictor')
plt.xlabel('Study Hours')
plt.ylabel('Marks')

plt.show()
```

```python
plt.scatter(df['Study_Hours'], df['Marks'])

plt.plot(df['Study_Hours'], model.predict(X))

plt.title('Student Marks Predictor')
plt.xlabel('Study Hours')
plt.ylabel('Marks')

plt.show()
```
